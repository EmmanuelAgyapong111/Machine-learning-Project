```python
# Separating the data and labels 
features = diabetes_dataset.drop(columns= 'Outcome', axis = 1)
target = diabetes_dataset['Outcome']
```

```python
# Data Standardization 

scaler = StandardScaler()
scaler.fit(features)
standardized_data = scaler.transform(features)
```

```python
print(standardized_data)
```

```python
features = standardized_data
target = diabetes_dataset["Outcome"]
```

```python
# Train Test Split 

X_train, X_test, Y_train, Y_test = train_test_split(features, target, test_size= 0.2, random_state=2)
```

```python
# Logistic Regression 

class Logistics_Regression():

    # declaring Learning rate and Number of Iteration (Hyperparameters)
    def __init__(self, Learning_rate, no_of_iteration):

        self.Learning_rate = Learning_rate
        self.no_of_iteration = no_of_iteration

    # Fit function to train the model with dataset
    def fit(self, X, Y):

        #[rows, input features], rows = m, input features = n
        self.m, self.n = X.shape 


        #initiating weight and bias values 
        self.w = np.zeros(self.n)

        self.b = 0

        self.X = X

        self.Y = Y

        # Implementing gradient descent for optimization 
        for i in range(self.no_of_iteration):
            self.update_weights()     


    def update_weights(self):

        # Y_hat formulat (Sigmoid function)
        Y_hat = 1 / (1 + np.exp( - (self.X.dot(self.w) + self.b)))

        # derivatives 
        dw = (1 / self.m)*np.dot(self.X.T, (Y_hat - self.Y))

        db = (1 / self.m)*np.sum(Y_hat - self.Y)

        # Updating the weight and bias using gradient descent equation 
        self.w = self.w - self.Learning_rate * dw 
        self.b = self.b - self.Learning_rate * db 

    # Sigmoid function and Decision making
    def predict(self, X):

        Y_pred = 1 / (1 + np.exp( - (X.dot(self.w) + self.b)))
        Y_pred = np.where(Y_pred > 0.5, 1, 0)
        return Y_pred
```

```python
classifier = Logistics_Regression(Learning_rate=0.01, no_of_iteration=1000)
```

```python
# Training the support vector Machine Classifier

classifier.fit(X_train, Y_train)
```

```python
# Model Evaluation 

# Accuracy Score 
# accuracy score on the training data 

X_train_prediction = classifier.predict(X_train)
training_data_accuracy = accuracy_score(Y_train, X_train_prediction)
```

```python
print('Accuracy score of the training data : ', training_data_accuracy)
```

```python
X_test_prediction = classifier.predict(X_test)
test_data_accuracy = accuracy_score(Y_test, X_test_prediction)
print('Accuracy score of the training data : ', test_data_accuracy)
```
