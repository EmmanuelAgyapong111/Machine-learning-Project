```python
#Handling missing values
#Methods to handle Missing values: 1. Imputation 2. Dropping 

import numpy as np
import pandas as pd 
import matplotlib.pyplot as plt 
import seaborn as sns 
```

```python
dataset = pd.read_csv("Placement_Dataset.csv")
dataset.head()
```

```python
dataset.info
```

```python
dataset.isnull().sum()
```

```python
#there are two ways to handle missing data 1. Imputation 2. Dropping 
# to use imputation, central tendencies are vital 1. Mean  2. Median   3. Mode 
```

```python
#Analyze the distribution of data in the column with missing values (Salary)
fig, ax = plt.subplots(figsize=(8,8))
sns.histplot(dataset.salary, stat='density', kde=True)
```

```python
#Replace the missing values with median values 
dataset["salary"].fillna(dataset["salary"].median(), inplace=True)
```

```python
dataset.isnull().sum()
```

```python
#dropping method 
salary_dataset = pd.read_csv("Placement_Dataset.csv")
```

```python
salary_dataset.shape
```

```python
salary_dataset.isnull().sum()
```

```python
#dropping the missing values 
salary_dataset = salary_dataset.dropna(how="any")
```

```python
salary_dataset.isnull().sum()
```
