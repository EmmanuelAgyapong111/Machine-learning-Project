```python
#This dataset is highly imbalanced
# 0 represent legit transactions and 1 represent fraudulent transactions
```

```python
#separating the legit and fraudulent transaction 
legit = credit_data[credit_data.Class == 0]
fraud = credit_data[credit_data.Class == 1]
```

```python
print(legit.shape)
print(fraud.shape)
```

```python
#There are several ways of handling imbalanced datasets. One way is Under-sampling 
#Under-sampling is done by building a sample dataset containing similar distribution of legit and fradulent transaction 
```

```python
#Now since we have 492 fraud data, our new under-sampling legit data must be similar or equal to 492 
legit_sample = legit.sample(n=492)
```

```python
#concatenate the two dataframes
new_dataset = pd.concat([legit_sample, fraud], axis=0)
```

```python
#Oversampling 
from sklearn.utils import resample 
```

```python
#separating the legit and fraudulent transaction 
Not_fraud = credit_data[credit_data.Class == 0]
Fraud = credit_data[credit_data.Class == 1]
```

```python
fraud_oversampled = resample(
    Fraud,
    n_samples=len(Not_fraud),
    random_state = 42
)
```

```python
New_dataset = pd.concat([Not_fraud, fraud_oversampled], axis = 0)
```

```python
#SMOTE (Synthetic Minority Oversampling Technique)
from imblearn.over_sampling import SMOTE
```

```python
X = credit_data.drop(columns='Class', axis = 1)
Y = credit_data['Class']

smote = SMOTE(random_state=42)
```

```python
print("Before SMOTE:", X.shape, Y.shape)
```

```python
X_resampled, Y_resampled = smote.fit_resample(X, Y)
```

```python
print("After SMOTE:", X_resampled.shape, Y_resampled.shape)
```

```python
print(Y_resampled.value_counts())
```
