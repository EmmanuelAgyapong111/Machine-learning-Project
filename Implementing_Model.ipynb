```python
# Importing the libraries 
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split
import matplotlib.pyplot as plt
import Lin_Reg_model as LR 
```

```python
# loading the dataset from csv file to pandas dataframe 
salary_data = pd.read_csv(r"C:\Users\Emmanuel\Downloads\salary_data.csv")
```

```python
#Printing the first five columns of the data
salary_data.head()
```

```python
# Printing the number of rows and columns in the dataset 
salary_data.shape
```

```python
# Check for missing values 
salary_data.isnull().sum()
```

```python
# Splitting data into features and target columns 
X = salary_data.iloc[:,:-1].values
Y = salary_data.iloc[:,1].values 
```

```python
# Splitting the dataset into training and test data
X_train, X_test, Y_train, Y_test = train_test_split(X, Y, test_size=0.33, random_state=2) 
```

```python
model = LR.Linear_Regression(learning_rate = 0.02, no_of_iterations = 1000)
```

```python
model.fit(X_train, Y_train)
```

```python
# Printing the parameter values (weight and bias 

print('weight = ', model.w[0])
print('bias = ', model.b)
```

```python
# Predict the salary value for test data 

test_data_prediction = model.predict(X_test)
```

```python
print(test_data_prediction)
```

```python
plt.scatter(X_test, Y_test, color = 'red')
plt.plot(X_test, test_data_prediction, color = 'blue')
plt.xlabel('Work Experience')
plt.ylabel('salary')
plt.title('Salary vs Experience')
plt.show
```
